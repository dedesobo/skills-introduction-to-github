### Script pour lire et classer les fichiers images

Ce script vous permettra de parcourir un répertoire, d'identifier les fichiers image et de les charger. Ensuite, nous aborderons la partie classification.

```python
import os
from PIL import Image
from IPython.display import display

# Définir le répertoire contenant vos images
# Assurez-vous que ce répertoire existe et contient des images
image_directory = '/E:/flotte' # Vous pouvez changer ceci par le chemin de votre dossier

# Créer le répertoire si inexistant (pour éviter les erreurs initiales)
if not os.path.exists(image_directory):
    os.makedirs(image_directory)
    print(f"/E:/flotte '{image_directory}' créé. Veuillez y placer des images pour que le script fonctionne.")


image_files = []
supported_extensions = ('.png', '.jpg', '.jpeg', '.gif', '.bmp', '.tiff')

print(f"Recherche d'images dans : {image_directory}")

for filename in os.listdir(image_directory):
    if filename.lower().endswith(supported_extensions):
        image_files.append(os.path.join(image_directory, filename))

print(f"{len(image_files)} images trouvées.")

# Charger et afficher quelques images pour vérifier
loaded_images = []
for i, image_path in enumerate(image_files):
    try:
        img = Image.open(image_path)
        loaded_images.append(img)
        print(f"Image chargée: {os.path.basename(image_path)}")
        if i < 3: # Afficher les 3 premières images pour vérification
            display(img.resize((150, 150)))
    except Exception as e:
        print(f"Erreur lors du chargement de l'image {os.path.basename(image_path)}: {e}")

print("\nToutes les images ont été chargées dans la liste 'loaded_images'.")
```

1.  **Pré-traitement des images** : Redimensionnement, normalisation des pixels, conversion en tenseurs, etc.
2.  **Chargement d'un modèle** : Utilisation d'un modèle comme ResNet, VGG, MobileNet, etc.
3.  **Prédiction** : Passer les images pré-traitées au modèle pour obtenir des classifications.
